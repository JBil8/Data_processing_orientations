```python
def read_pdf_theta(directory, fixed_cof, fixed_I, aspect_ratios, direction):
    
    os.chdir(directory)
    #initilize data list
    actual_data = []
    # Loop over the aspect ratios (ap)
    for idx, ap in enumerate(aspect_ratios):
        filename = f'simple_shear_ap{ap}_cof_{fixed_cof}_I_{fixed_I}.pkl'
        if os.path.exists(filename):
            with open(filename, 'rb') as file:
                file_data = pickle.load(file)
                if direction == 'x':
                    pdf_data = file_data.get('pdf_thetax', None)
                elif direction == 'z':
                    pdf_data = file_data.get('pdf_thetaz', None)
                else:
                    raise ValueError("Invalid direction. Choose either 'x' or 'z'.")

                if pdf_data is not None and pdf_data.shape == (180, 2):
                    pdf_values = pdf_data[:, 1]  # The PDF values
                    bin_centers = pdf_data[:, 0]  # The bin centers
                    if ap<1:
                        bin_centers = bin_centers + np.pi/2
                        bin_centers = np.where(bin_centers>np.pi/2, bin_centers-np.pi, bin_centers)
                          # Sort the data to avoid crossing boundaries
                        sorted_indices = np.argsort(bin_centers)
                        bin_centers = bin_centers[sorted_indices]
                        pdf_values = pdf_values[sorted_indices]
                    reshaped_tuple = tuple((bin_centers, pdf_values))
                    actual_data.append(reshaped_tuple)

    return actual_data



def read_data_key(key, directory, fixed_cof, fixed_I, aspect_ratios):
    """
    Read the data from the pickle files with the given key.
    """
    os.chdir(directory)
    # Initialize the data list
    actual_data = []
    # Loop over the aspect ratios (ap)
    for idx, ap in enumerate(aspect_ratios):
        filename = f'simple_shear_ap{ap}_cof_{fixed_cof}_I_{fixed_I}.pkl'
        if os.path.exists(filename):
            with open(filename, 'rb') as file:
                file_data = pickle.load(file)
                # Extract the data
                data = file_data.get(key, None)
                if data is not None:
                    actual_data.append(data)
    return actual_data
```

```python
import numpy as np
from scipy.optimize import minimize
from scipy.interpolate import interp2d

def maier_saupe_fit_function(theta, phi, r, phi_n):
    """
    Evaluate the Maier–Saupe-like PDF over theta, phi.
    theta: polar angle (radians)
    phi: azimuthal angle (radians)
    r: strength parameter
    phi_n: azimuthal angle of director in shear plane
    """
    sin_theta = np.sin(theta)
    cos_phi_shifted = np.cos(phi - phi_n)
    
    argument = (3/4) * r * (1.5 * sin_theta**2 * cos_phi_shifted**2 - 1)
    return np.exp(argument)

def normalize_pdf(pdf, theta, phi):
    """ Normalize over theta and phi using spherical integration weight """
    dtheta = np.abs(theta[1] - theta[0])
    dphi = np.abs(phi[1] - phi[0])
    weight = np.sin(theta[:, None])  # shape (N, 1)
    norm = np.sum(pdf * weight) * dtheta * dphi
    return pdf / norm

def loss_function(params, theta_grid, phi_grid, target_pdf):
    r, phi_n = params
    model = maier_saupe_fit_function(theta_grid, phi_grid, r, phi_n)
    model_normalized = normalize_pdf(model, theta_grid[:,0], phi_grid[0,:])
    return np.sum((model_normalized - target_pdf)**2)

def fit_maier_saupe_pdf(theta_edges, phi_edges, pdf_data):
    """
    Fit the PDF data using the Maier–Saupe 3D form.
    Inputs:
        - theta_edges, phi_edges: 1D bin edges in radians
        - pdf_data: 2D histogram of PDF values (shape: [len(theta)-1, len(phi)-1])
    Returns:
        - r_fit, phi_n_fit, loss
    """
    # Get bin centers
    theta_centers = 0.5 * (theta_edges[:-1] + theta_edges[1:])
    phi_centers = 0.5 * (phi_edges[:-1] + phi_edges[1:])
    theta_grid, phi_grid = np.meshgrid(theta_centers, phi_centers, indexing='ij')

    # Optimization
    initial_guess = [2.0, 0.0]  # r and phi_n
    bounds = [(0, 30), (-np.pi/2, np.pi/2)]

    result = minimize(
        loss_function,
        initial_guess,
        # args=(theta_grid, phi_grid, pdf_data),
        args=(phi_grid, theta_grid, pdf_data),
        bounds=bounds
    )

    r_fit, phi_n_fit = result.x
    return r_fit, phi_n_fit, result.fun
```

```python
def maier_saupe_curve_fit_model(X, r, phi_n):
    """
    X: stacked array of shape (2, N), where
       X[0, :] = theta.ravel()
       X[1, :] = phi.ravel()
    Returns:
       1D array of model values, same shape as theta.ravel()
    """
    theta = X[0]
    phi = X[1]
    
    sin_theta = np.sin(theta)
    cos_phi_shifted = np.cos(phi - phi_n)
    argument = (3/4) * r * (1.5 * sin_theta**2 * cos_phi_shifted**2 - 1)
    
    # Compute unnormalized model
    unnormalized = np.exp(argument)
    
    # Normalize it
    # Approximate integration weights for normalization
    # Assume theta and phi grids were uniform
    dtheta = np.abs(np.unique(theta)[1] - np.unique(theta)[0])
    dphi = np.abs(np.unique(phi)[1] - np.unique(phi)[0])
    weight = np.sin(theta)
    norm = np.sum(unnormalized * weight) * dtheta * dphi
    
    return (unnormalized / norm)
```
