```python
from sympy import Znm, Symbol, simplify, expand, re, im, Ynm, sqrt, factor
from sympy.abc import n, m, p, q, U, beta, l
from sympy import Matrix, MatrixSymbol
from sympy import integrate, sin, pi
from sympy import Function
theta = Symbol('theta', positve=True, real=True)
phi = Symbol('phi', positve=True, real=True)
Znm(n, m, theta, phi)
```

```python
simplify(Znm(2, -2, theta, phi)).expand()   # I check that this is true
```

```python
basis = []
indices = []
L_max = 1
for l in range(L_max + 1):
    for m in range(-l, l+1):
        basis.append(Znm(l, m, theta, phi))
        indices.append((l, m))

# Create unknown coefficients a_lm
a = [Function(f'a_{l}_{m}')() for (l, m) in indices]

# Build f1 as sum of real spherical harmonics
f1 = sum(ai * Yi for ai, Yi in zip(a, basis))
```

```python
Uex = 0
for j, (l, m) in enumerate(indices):
    if l == 2:
        Uex += a[j] * basis[j]  # this is symbolic, doesn't integrate numerically yet
```

```python
laplace_beltrami = lambda l, m: -l*(l+1)

rhs_terms = []
for j, (l, m) in enumerate(indices):
    Δf1_Y = laplace_beltrami(l, m) * basis[j]
    # Laplacian of Uex * f1 is product rule:
    rhs_terms.append(a[j] * Δf1_Y)
```

```python
A = Matrix.zeros(len(indices), len(indices))
b = Matrix.zeros(len(indices), 1)

for i, Yi in enumerate(basis):
    for j, Yj in enumerate(basis):
        integrand = Yi * rhs_terms[j] * sin(theta)
        A[i, j] = simplify(integrate(
            integrate(integrand, (phi, 0, 2*pi)),
            (theta, 0, pi)
        ))
    b[i] = 0  # LHS projection, assuming f0 divergence term known separately
```
